### API (Application programming interface )
###### Api is a data pipeline which makes two software to talk to each other.Using tmdb movie database,tmdb my api is 'b13e21af3fc75883e78154a70c261f36' . and i want top rated movie so searching tmdb movie toprated api and then getting that api link and and than pasting my api link in between and than when i run i get json file."https://api.themoviedb.org/3/movie/top_rated?api_key=b13e21af3fc75883e78154a70c261f36&language=en-US&page=1"

we copy paste entire json data and paste on json viewer website to make data readable and from there we can select columns that we need and entire data will be in dictanory format 

```python
import pandas as pd
import requests 
```

```python
response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=b13e21af3fc75883e78154a70c261f36&language=en-US&page=1")
```

```python
movie_df=pd.DataFrame(response.json()['results'])
```

```python
movie_df[["id","title","overview","release_date","popularity","vote_average","vote_count"]]
```

```python
df = pd.DataFrame()
```

Above what we did gave us page 1 movies in total 20 ,so now we have 490 such pages and to get all the movies we will extract through loop.

```python
for i in range(1,491):
    response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=b13e21af3fc75883e78154a70c261f36&language=en-US&page={}".format(i))
    temp_df = pd.DataFrame(response.json()['results'])[["id","title","overview","release_date","popularity","vote_average","vote_count"]]
    df = pd.concat([df,temp_df],ignore_index=True)
```

```python
df.shape
```

```python
df.to_csv("movies.csv")
```
