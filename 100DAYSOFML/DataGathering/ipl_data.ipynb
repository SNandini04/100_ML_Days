```python
import requests
from bs4 import BeautifulSoup
import csv

# List of URLs to scrape
urls = [
    "https://www.espncricinfo.com/records/season/team-match-results/2007to08-2007to08?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2009-2009?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2009to10-2009to10?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2011-2011?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2012-2012?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2013-2013?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2014-2014?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2015-2015?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2016-2016?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2017-2017?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2018-2018?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2019-2019?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2020to21-2020to21?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2021-2021?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2022-2022?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2023-2023?trophy=117",
    "https://www.espncricinfo.com/records/season/team-match-results/2024-2024?trophy=117",
]

# CSV file to save the data
output_file = "ipl_match_results_all_seasons.csv"

# Open the CSV file for writing
with open(output_file, "w", newline="", encoding="utf-8") as csvfile:
    csvwriter = csv.writer(csvfile)
    first_table = True  # To write headers only once

    # Iterate through each URL
    for url in urls:
        print(f"Scraping: {url}")
        response = requests.get(url)
        soup = BeautifulSoup(response.content, 'html.parser')

        # Find the table containing match results
        table = soup.find('table')
        if not table:
            print(f"No table found on page: {url}")
            continue

        # Extract headers and rows
        headers = [th.text.strip() for th in table.find_all('th')]
        rows = [
            [td.text.strip() for td in tr.find_all(['td', 'th'])]
            for tr in table.find_all('tr') if tr.find_all(['td', 'th'])
        ]

        # Write headers only for the first table
        if first_table and headers:
            csvwriter.writerow(headers)
            first_table = False

        # Write rows to the CSV file
        csvwriter.writerows(rows)

print(f"Data successfully saved to '{output_file}'.")
```
