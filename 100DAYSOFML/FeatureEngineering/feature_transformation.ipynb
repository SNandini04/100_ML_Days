Missing Values Imputation , Handling categorical values , Outlier detection , Feature Scaling 

Feature Transform - Encoding 
1. Label Encoder is specifically used for target variables which is categorical .for example spam classifier target variable(y) - yes/no
2. Ordinal encoder is used for input variables (x) where x variable is ordinal categorical data .Ordinal means order .such as reviews Good ,better , worst . we can see order that is good > better > worst
3. One Hot encoder  is used where input categorical variable is nominal that is not ordered.such as states , engineering branches we can't say that state is better than this state 

### Ordinal Encoding

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\karth\Downloads\customer.csv")
```

```python
df.sample(5)
```

#### Gender is nominal data there is no order just category male and female . 'Nominal Encoder'
#### Review ,education is a ordinal data 'Ordinal Encoder'
#### Purchased is our target variable 'Label Encoder'

```python
# as of now we are ignoring gender n age 
df= df.iloc[:,2:]
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.2)
```

```python
X_train
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
## I am just giving categories in a order from low value to high value so good will get high value than poor and average
##  If i will not give categories randomly it will provide values 
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
oe.fit(X_train)
```

```python
X_train = oe.transform(X_train)
X_test = oe.transform(X_test)
```

```python
X_train
```

```python
oe.categories_
```

## OneHotEncoding with Top Categories

```python
counts = df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold = 100
```

```python
# we are just saying cars less than 100 in value counts storing its index (name of brand) in repl 
repl=counts[counts<=threshold].index
```

```python
# Whereever there is repl we put into ucommon column 
pd.get_dummies(df['brand'].replace(repl,'uncommon')).sample(10)
```
